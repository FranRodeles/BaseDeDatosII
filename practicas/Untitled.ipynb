```python
# Ejecutá esta celda para verificar tu entorno
# No es necesario modificarla

import sys
import importlib

libs = {
'numpy': '1.24',
'pandas': '2.0',
'seaborn': '0.12',
'matplotlib': '3.7',
}

print(f"Python: {sys.version.split()[0]}")
print("-" * 40)

all_ok = True
for lib, min_ver in libs.items():
    try:
        mod = importlib.import_module(lib)
        ver = mod.version
        status = "✅" if ver >= min_ver else "⚠️ versión antigua"
        print(f"{lib:15s} {ver:10s} {status}")
    except ImportError:
        print(f"{lib:15s} {'NO INSTALADA':10s} ❌")
        all_ok = False

print("-" * 40)
print("Entorno OK ✅" if all_ok else "Revisá las dependencias faltantes ❌")
```

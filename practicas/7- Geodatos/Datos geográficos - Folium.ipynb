## Marcadores

```python
# Crear mapa
sr_map = folium.Map(location=[-34.6134945536014, -68.32745300275877], zoom_start=30)
#Marcador simple
marker = folium.Marker(
    location=[-34.6134945536014, -68.32745300275877])
    # Agregar marcador a mapa
marker.add_to(sr_map)

#Marcador con ícono
marker = folium.Marker(
    location=[-34.6134945536014, -68.327000000000],
    icon=folium.Icon(icon="home"))
# Agregar marcador a mapa
marker.add_to(sr_map)

#Marcador con ícono - color
marker = folium.Marker(
    location=[-34.6134945536014, -68.326000000000],
    icon=folium.Icon(icon="home",color="red",icon_color="black")
    )
# Agregar marcador a mapa
marker.add_to(sr_map)

#Marcador con ícono y texto
marker = folium.Marker(
    location=[-34.6134945536014, -68.328000000000],
    tooltip="Leyenda",
    icon=folium.Icon(icon="trash"))
# Agregar marcador a mapa
marker.add_to(sr_map)

sr_map
```

## Marcador circular

```python
sr_map = folium.Map(location=[-34.6134945536014, -68.32745300275877], zoom_start=30)
# Crear marcador

#Marcador circular
marker = folium.CircleMarker(
    location=[-34.6134945536014, -68.32745300275877])
    # Agregar marcador a mapa
marker.add_to(sr_map)

#Marcador circular
marker = folium.CircleMarker(
    location=[-34.6134945536014, -68.327000000000],
     radius = 15)
# Agregar marcador a mapa
marker.add_to(sr_map)

#Marcador circular
marker = folium.CircleMarker(
    location=[-34.6134945536014, -68.327000000000],
    radius = 30)
# Agregar marcador a mapa
marker.add_to(sr_map)

sr_map
```

## Heatmaps

```python
from folium import plugins

# Crear Mapa
heat_m = folium.Map([51.509443, 0.051117], zoom_start=10)

import pandas as pd
# Leer csv
df = pd.read_csv('../data/incident_2017.csv', decimal=",")
df = df[["Latitude","Longitude"]]
df = df.dropna()
df
```

```python
values = df.values
# Agregar Heat Map
heat_m.add_child(plugins.HeatMap(values, radius=14))
heat_m
```

```python
# Crear mapa
heat_m_2 = folium.Map([51.509443, 0.051117], zoom_start=10)
# Agregar mapa de calor como grupo
plugins.HeatMap(values).add_to(folium.FeatureGroup(name='Heat Map').add_to(heat_m_2))
# Agregar capa de control
folium.LayerControl().add_to(heat_m_2)
# Mostrar mapa
heat_m_2
```

## Capas

```python
tiles = ["OpenStreetMap","Stamen Toner", "Stamen Watercolor","CartoDB positron", "CartoDB dark_matter"]
# Crear mapa
layer_map = folium.Map(location=[-34.6134945536014, -68.32745300275877], titles=tiles[2], zoom_start=15)
layer_map
```

```python
# Crear mapa
m_layers = folium.Map(location=[-34.6134945536014, -68.32745300275877], zoom_start=15)
# Cargar capas
folium.TileLayer('Stamen Toner').add_to(m_layers)
folium.TileLayer('Stamen Watercolor').add_to(m_layers)
folium.TileLayer('CartoDB positron').add_to(m_layers)
folium.TileLayer('CartoDB dark_matter').add_to(m_layers)
# Crear capa de control
folium.LayerControl().add_to(m_layers)
m_layers
```

## Ejercicio 1

- Descomprimir el archivo clima.tgz de la carpeta data
- Convertir cada archivo a Dataframe y mostrarlo en pantalla

```python
import tarfile
import os
import pandas as pd

# 1. Descomprimir el archivo clima.tgz de la carpeta data
tar_path = "../data/clima.tgz"
extract_path = "../data/clima-tp1ejer1"

os.makedirs(extract_path, exist_ok=True)
# Nota: clima.tgz es un archivo tar sin compresion gzip adicional, por lo que se abre con "r:*"
with tarfile.open(tar_path, "r:*") as tar:
    tar.extractall(path=extract_path)

print(f"Archivos extraidos correctamente en: {extract_path}")

# 2. Convertir cada archivo a Dataframe y mostrarlo en pantalla
archivos = sorted([f for f in os.listdir(extract_path) if f.endswith(".txt")])
dataframes = {}

for archivo in archivos:
    ruta_archivo = os.path.join(extract_path, archivo)
    df = pd.read_csv(ruta_archivo, delimiter="\t", encoding="latin1")
    dataframes[archivo] = df
    print(f"=== Dataframe: {archivo} (Filas: {df.shape[0]}, Columnas: {df.shape[1]}) ===")
    display(df.head())
```

## Ejercicio 2

- Obtener de la API de <a href="https://wikimedia.org/api/rest_v1/#/Pageviews%20data">Wikimedia </a> la cantidad de visitas en este año a la página Albert Einstein por día.
- Mostrar el Dataframe obtenido por pantalla

```python
import requests
import datetime
import pandas as pd

# 1. Parametros para la API de Wikimedia
articulo = "Albert_Einstein"
proyecto = "es.wikipedia"  # Wikipedia en espanol
granularity = "daily"

# Obtener fechas del anio en curso
anio_actual = datetime.datetime.now().year
fecha_inicio = f"{anio_actual}0101"
fecha_fin = datetime.datetime.now().strftime("%Y%m%d")

url_wikimedia = f"https://wikimedia.org/api/rest_v1/metrics/pageviews/per-article/{proyecto}/all-access/all-agents/{articulo}/{granularity}/{fecha_inicio}/{fecha_fin}"

# Wikimedia requiere un User-Agent identificatorio para evitar error 403
headers = {
    "User-Agent": "TP1_RepositoriosInformacion/1.0 (contacto@estudiante.um.edu.ar)"
}

# 2. Realizar peticion y procesar respuesta
response = requests.get(url_wikimedia, headers=headers)

if response.status_code == 200:
    datos = response.json().get("items", [])
    df_wikimedia = pd.DataFrame(datos)
    
    # Formatear la columna de fecha
    df_wikimedia["fecha"] = pd.to_datetime(df_wikimedia["timestamp"].str.slice(0, 8), format="%Y%m%d")
    df_resultado = df_wikimedia[["fecha", "article", "granularity", "views"]]
    
    print(f"Cantidad de registros diarios obtenidos para {articulo} en {anio_actual}: {len(df_resultado)}")
    display(df_resultado)
else:
    print(f"Error al consultar la API: {response.status_code} - {response.text}")
```

## Ejercicio 3

- Crear una base de datos MySQL
- Crear una tabla que contenga nombre, apellido y teléfono
- Obtener de la base de datos la lista de personas ordenada por apellido de forma descendente
- Mostrar el Dataframe resultante por pantalla

```python
import mysql.connector as connection
import pandas as pd
import warnings

# Suprimir advertencias informativas de pandas sobre conexiones DBAPI2 directas
warnings.filterwarnings("ignore")

# 1. Conexion al servidor MySQL local
db = connection.connect(
    host="localhost",
    user="root",
    passwd="root"
)
cursor = db.cursor()

# 2. Crear una base de datos MySQL si no existe
cursor.execute("CREATE DATABASE IF NOT EXISTS tp1_db;")
cursor.execute("USE tp1_db;")

# 3. Crear una tabla que contenga nombre, apellido y telefono
cursor.execute("""
CREATE TABLE IF NOT EXISTS personas (
    id INT AUTO_INCREMENT PRIMARY KEY,
    nombre VARCHAR(50) NOT NULL,
    apellido VARCHAR(50) NOT NULL,
    telefono VARCHAR(20) NOT NULL
);
""")

# Insertar registros de prueba si la tabla esta vacia
cursor.execute("SELECT COUNT(*) FROM personas;")
if cursor.fetchone()[0] == 0:
    datos_prueba = [
        ("Juan", "Pérez", "261-4567890"),
        ("María", "Gómez", "261-5678901"),
        ("Carlos", "Álvarez", "261-6789012"),
        ("Ana", "Zárate", "261-7890123"),
        ("Lucía", "Fernández", "261-8901234")
    ]
    cursor.executemany("INSERT INTO personas (nombre, apellido, telefono) VALUES (%s, %s, %s);", datos_prueba)
    db.commit()

# 4. Obtener de la base de datos la lista de personas ordenada por apellido de forma descendente
query = "SELECT nombre, apellido, telefono FROM personas ORDER BY apellido DESC;"
df_personas = pd.read_sql(query, db)

# Cerrar cursor y conexion
cursor.close()
db.close()

# 5. Mostrar el Dataframe resultante por pantalla
print("=== Lista de personas ordenada por apellido descendente ===")
display(df_personas)
```

## Ejercicio 4

- Obtener de la API <a href="https://date.nager.at/swagger/index.html">Nager.Date</a> las fiestas nacionales argentinas para el 2021
- Mostrar el dataframe resultante por pantalla

```python
import requests
import pandas as pd

# 1. Obtener de la API Nager.Date las fiestas nacionales argentinas para el 2021
url_nager = "https://date.nager.at/api/v3/PublicHolidays/2021/AR"
response = requests.get(url_nager)

if response.status_code == 200:
    datos_feriados = response.json()
    
    # 2. Convertir respuesta JSON a Dataframe
    df_feriados = pd.DataFrame(datos_feriados)
    
    # Seleccionar y mostrar columnas principales
    columnas_interes = ["date", "localName", "name", "countryCode", "types"]
    df_feriados_res = df_feriados[columnas_interes]
    
    print(f"Total de fiestas nacionales en Argentina (2021): {len(df_feriados_res)}")
    display(df_feriados_res)
else:
    print(f"Error al consultar la API Nager.Date: {response.status_code} - {response.text}")
```

## Ejercicio 5

- Obtener una tabla informativa de una página a elección con Web Scrapping
- Mostrar el Dataframe resultante por pantalla

```python
import requests
import pandas as pd
from io import StringIO

# 1. Obtener una tabla informativa con Web Scraping
# Tabla de paises de America del Sur por poblacion desde Wikipedia
url_scraping = "https://es.wikipedia.org/wiki/Anexo:Pa%C3%ADses_de_Am%C3%A9rica_del_Sur_por_poblaci%C3%B3n"
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36"
}

response = requests.get(url_scraping, headers=headers)

if response.status_code == 200:
    # Leer tablas HTML usando pandas
    tablas = pd.read_html(StringIO(response.text))
    
    # Seleccionar la tabla de paises (segunda tabla encontrada en la pagina)
    df_paises = tablas[1]
    
    # Limpiar columna de referencias si existe
    if "Ref." in df_paises.columns:
        df_paises = df_paises.drop(columns=["Ref."])
        
    print("=== Tabla informativa obtenida por Web Scraping: Paises de Sudamerica por Poblacion ===")
    display(df_paises)
else:
    print(f"Error al realizar Web Scraping: {response.status_code}")
```
